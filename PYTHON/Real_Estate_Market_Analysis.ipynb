````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
````

````python
df = pd.read_csv("../EXCEL/real_estate_cleaned.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
````

````python
import os

print(os.getcwd())
print(os.listdir())
````

````python
# Basic Dataset Validation

print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
print("\nData Types:")
print(df.dtypes)
````

````python
# Check Missing Values

missing_values = df.isnull().sum()
print("Missing Values by Columns:")
print(missing_values[missing_values >0])
````

````python
# Check duplicate records

print("Duplicate Rows:", df.duplicated().sum())
print("Duplicate Property IDs:", df["Property_ID"].duplicated().sum())
````

````python
# Price Distribution Analysis

plt.Figure(figsize=(10, 6))
sns.histplot(
    df["Price_INR_Lakhs"],
    bins= 40,
    kde=True
)
plt.title("Distribution of Property Prices")
plt.xlabel("Property Price (₹ Lakhs)")
plt.ylabel("Number of Properties")
plt.show()
````

````python
plt.Figure(figsize=(10, 6))
sns.histplot(
    df["Price_Per_SqFt"],
    bins=40,
    kde=True
)
plt.title("Distribution of Property Price per Sq.Ft.")
plt.xlabel("Price per Sq.Ft. (₹)")
plt.ylabel("Number of Properties")
plt.show()
````

````python
# Property Count by City

city_counts = df["City"].value_counts()

print(city_counts)
````

````python
plt.figure(figsize=(9, 6))

city_counts.plot(kind="bar")

plt.title("Property Inventory by City")
plt.xlabel("City")
plt.ylabel("Number of Properties")
plt.xticks(rotation=0)
plt.show()
````

````python
# Average Property Price by City

city_avg_price = (
    df.groupby("City")["Price_INR_Lakhs"]
    .mean()
    .sort_values(ascending=False)
)

print(city_avg_price)
````

````python
plt.figure(figsize=(9, 6))

city_avg_price.plot(kind="bar")

plt.title("Average Property Price by City")
plt.xlabel("City")
plt.ylabel("Average Price (₹ Lakhs)")
plt.xticks(rotation=0)
plt.show()
````

````python
# Average Price per Sq.Ft. by City

city_avg_psf = (
    df.groupby("City")["Price_Per_SqFt"]
    .mean()
    .sort_values(ascending=False)
)

print(city_avg_psf)
````

````python
plt.figure(figsize=(9, 6))

city_avg_psf.plot(kind="bar")

plt.title("Average Price per Sq.Ft. by City")
plt.xlabel("City")
plt.ylabel("Average Price per Sq.Ft. (₹)")
plt.xticks(rotation=0)
plt.show()
````

````python
# Property Distribution by BHK

bhk_counts = df["BHK"].value_counts().sort_index()

print(bhk_counts)
````

````python
plt.figure(figsize=(9, 6))

bhk_counts.plot(kind="bar")

plt.title("Property Distribution by BHK")
plt.xlabel("BHK")
plt.ylabel("Number of Properties")
plt.xticks(rotation=0)
plt.show()
````

````python
# Average Property Price by BHK

bhk_avg_price = (
    df.groupby("BHK")["Price_INR_Lakhs"]
    .mean()
    .sort_index()
)

print(bhk_avg_price)
````

````python
plt.figure(figsize=(9, 6))

bhk_avg_price.plot(kind="bar")

plt.title("Average Property Price by BHK")
plt.xlabel("BHK")
plt.ylabel("Average Price (₹ Lakhs)")
plt.xticks(rotation=0)
plt.show()
````

````python
# Property Type Distribution

property_type_counts = df["Property_Type"].value_counts()

print(property_type_counts)
````

````python
plt.figure(figsize=(9, 6))

property_type_counts.plot(kind="bar")

plt.title("Property Distribution by Property Type")
plt.xlabel("Property Type")
plt.ylabel("Number of Properties")
plt.xticks(rotation=0)
plt.show()
````

````python
# Average Price by Property Type

property_type_avg_price = (
    df.groupby("Property_Type")["Price_INR_Lakhs"]
    .mean()
    .sort_values(ascending=False)
)

print(property_type_avg_price)
````

````python
plt.figure(figsize=(9, 6))

property_type_avg_price.plot(kind="bar")

plt.title("Average Property Price by Property Type")
plt.xlabel("Property Type")
plt.ylabel("Average Price (₹ Lakhs)")
plt.xticks(rotation=0)
plt.show()
````

````python
# Area vs Property Price

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Area_SqFt",
    y="Price_INR_Lakhs",
    hue="Property_Type",
    alpha=0.6
)

plt.title("Property Area vs Price")
plt.xlabel("Area (Sq.Ft.)")
plt.ylabel("Property Price (₹ Lakhs)")
plt.legend(title="Property Type")
plt.show()
````

````python
# Correlation between Area and Price

area_price_corr = df["Area_SqFt"].corr(df["Price_INR_Lakhs"])

print("Correlation between Area and Price:", round(area_price_corr, 3))
````

````python
# Area vs Price per Sq.Ft.

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Area_SqFt",
    y="Price_Per_SqFt",
    hue="City",
    alpha=0.6
)

plt.title("Property Area vs Price per Sq.Ft.")
plt.xlabel("Area (Sq.Ft.)")
plt.ylabel("Price per Sq.Ft. (₹)")
plt.legend(title="City")
plt.show()
````

````python
# Property Age vs Average Price

age_price = (
    df.groupby("Property_Age_Years")["Price_INR_Lakhs"]
    .mean()
    .sort_index()
)

print(age_price)
````

````python
plt.figure(figsize=(10, 6))

age_price.plot(kind="line", marker="o")

plt.title("Property Age vs Average Property Price")
plt.xlabel("Property Age (Years)")
plt.ylabel("Average Price (₹ Lakhs)")
plt.grid(True)
plt.show()
````

````python
# Create Property Age Groups

df["Property_Age_Group"] = pd.cut(
    df["Property_Age_Years"],
    bins=[-1, 5, 10, 20, float("inf")],
    labels=["0-5 Years", "6-10 Years", "11-20 Years", "20+ Years"]
)

age_group_analysis = (
    df.groupby("Property_Age_Group", observed=True)["Price_INR_Lakhs"]
    .mean()
)

print(age_group_analysis)
````

````python
plt.figure(figsize=(9, 6))

age_group_analysis.plot(kind="bar")

plt.title("Average Property Price by Property Age Group")
plt.xlabel("Property Age Group")
plt.ylabel("Average Price (₹ Lakhs)")
plt.xticks(rotation=0)
plt.show()
````

````python
# Metro Distance vs Price per Sq.Ft.

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Distance_to_Metro_km",
    y="Price_Per_SqFt",
    hue="City",
    alpha=0.6
)

plt.title("Metro Distance vs Price per Sq.Ft.")
plt.xlabel("Distance to Metro (km)")
plt.ylabel("Price per Sq.Ft. (₹)")
plt.legend(title="City")
plt.show()
````

````python
# Correlation between Metro Distance and Price per Sq.Ft.

metro_price_corr = df["Distance_to_Metro_km"].corr(
    df["Price_Per_SqFt"]
)

print(
    "Correlation between Metro Distance and Price per Sq.Ft.:",
    round(metro_price_corr, 3)
)
````

````python
# Final Market Summary

print("REAL ESTATE MARKET SUMMARY")
print("-" * 40)

print("Total Properties:", len(df))
print("Average Price (₹ Lakhs):", round(df["Price_INR_Lakhs"].mean(), 2))
print("Median Price (₹ Lakhs):", round(df["Price_INR_Lakhs"].median(), 2))
print("Average Price per Sq.Ft. (₹):", round(df["Price_Per_SqFt"].mean(), 2))
print("Average Area (Sq.Ft.):", round(df["Area_SqFt"].mean(), 2))

print("\nProperties by City:")
print(df["City"].value_counts())

print("\nMost Common Property Type:")
print(df["Property_Type"].mode()[0])

print("\nMost Common BHK:")
print(df["BHK"].mode()[0])
````
